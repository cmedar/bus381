# Bronze — Raw Data Profile

Loads the three raw CSVs from the EC2 data volume and profiles them.
No transformation — this layer is read-only.

Set `DATA_DIR` env var to the folder containing your local copy of the EC2 data,
or place the CSV files in `../../data/` relative to this notebook.

```python
import os
import pandas as pd
from pathlib import Path

DATA_DIR = Path(os.getenv("DATA_DIR", "../../data"))
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.1f}".format)
```

## journeys.csv

```python
journeys = pd.read_csv(DATA_DIR / "journeys.csv")
print(f"Rows: {len(journeys):,}  |  Columns: {len(journeys.columns)}")
print(f"Date range: {journeys['sincai_at'].min()} → {journeys['sincai_at'].max()}")
journeys.dtypes
```

```python
print("Null counts:")
print(journeys.isnull().sum())
print("\ntotal_seconds stats:")
print(pd.to_numeric(journeys["total_seconds"], errors="coerce").describe())
```

```python
journeys.head(3)
```

## crossings.csv

```python
crossings = pd.read_csv(DATA_DIR / "crossings.csv")
print(f"Rows: {len(crossings):,}  |  Columns: {len(crossings.columns)}")
print(f"Date range: {crossings['crossed_at'].min()} → {crossings['crossed_at'].max()}")
print(f"\nCrossings per stop (dir0):")
print(crossings[crossings['direction']==0].groupby('stop_name').size().sort_values(ascending=False))
```

```python
crossings.head(3)
```

## arrivals.csv

```python
arrivals = pd.read_csv(DATA_DIR / "arrivals.csv")
print(f"Rows: {len(arrivals):,}  |  Columns: {len(arrivals.columns)}")
print(f"Date range: {arrivals['ingested_at'].min()} → {arrivals['ingested_at'].max()}")
print(f"\nLive vs scheduled readings:")
print(arrivals['is_timetable'].value_counts())
```
